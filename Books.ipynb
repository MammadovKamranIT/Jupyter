```python
import pandas as pd


data = {
    "Title": [
        "Harry Potter and the Philosopher's Stone",
        "The Hobbit",
        "1984",
        "Pride and Prejudice",
        "The Great Gatsby",
        "To Kill a Mockingbird"
    ],
    "Author": [
        "J.K. Rowling",
        "J.R.R. Tolkien",
        "George Orwell",
        "Jane Austen",
        "F. Scott Fitzgerald",
        "Harper Lee"
    ],
    "Year of publication": [
        1997, 1937, 1949, 1813, 2025, 2016
    ],
    "Price": [
        15.99, 12.50, 10.99, 9.50, 11.25, 13.75
    ]
}

df = pd.DataFrame(data)

df
```

```python
average_price = df['Price'].mean()
average_price
```

```python
BooksAfter2015 = df[df['Year of publication'] > 2015]
BooksAfter2015
```

```python
sortedDf = df.sort_values(by='Price')
sortedDf
```
